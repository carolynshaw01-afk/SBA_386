```python
# Lesson 386.6 - Matplotlib Line Plot
# # Lesson 386.5 - Getting Started with Data Visualization Using Python, pages 20-22
# R-GLAB 386.5.1 - Plotting Line Styles
# 🧠 Carolyn's Self-Talk:
# "plt.plot tells me LINE chart."
# "month_number gives me my X-axis."
# "total_profit gives me my Y-axis."
# "Connect my monthly profit points with a line."
plt.plot(df['month_number'], df['total_profit'])
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
```

# Phase 3 → Meet every SBA requirement

```python
# Lesson 386.6 - Matplotlib Line Plot
# # Lesson 386.5 - Getting Started with Data Visualization Using Python, pages 29
# R-GLAB 386.5.1 - Plotting Line Styles
# 🧠 Carolyn's Self-Talk:
# "plt.plot tells me LINE chart."
# "month_number gives me my X-axis."
# "total_profit gives me my Y-axis."
# "Connect my monthly profit points with a line."
plt.plot(df['month_number'], df['total_profit'],
         linestyle=':',
         color='red')

plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
# 🧠 Carolyn's translation
# Read that sucker left to right:
#plt.plot → “Make me a line chart.”
# df['month_number'] → “These are my X values.”
# df['total_profit'] → “These are my Y values.”
# linestyle=':' → “Make my line dotted.”
# color='red' → “Make my line red.”
```

# Optional Phase 4 → Carolyn makes it pretty

```python
plt.plot(df['month_number'], df['total_profit'],
         linestyle=':',
         color='red',
         marker='o',
         markerfacecolor='black',
         linewidth=3,
         label='Profit data of last year')
plt.legend(loc='lower right')
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
# Caroln's Self-Talk:
# marker='o' → “Put a circle on each data point.”
# 🧠 Why o?
# Think, think, think:
# o looks like a little circle. → ○
# markerfacecolor='black' → “Make the inside of my circles black.”
# linewidth=3 → “Make my red dotted line thicker.”
# Add labels, titles, and legends for readability.  Page 4, 11, 16-Lesson 386.5-Getting Started with Data Visualization Using Python
# label= -> "What NAME should appear in my legend?"
# "Call this line Profit data of last year."
# plt.legend -> "Show me the legend."
# loc -> "WHERE do I want it?"
# 'lower right' -> "Put it in the bottom-right corner."
# Read it left → right:
# plt → go into Matplotlib → .legend() → give me a legend → loc= → put it WHERE? → 'lower right'.
# My graph currently has:
# dotted ✅ → red ✅ → circles ✅ → black marker faces ✅ → linewidth 3 ✅ → label ✅ → legend ✅
# X label ✅ → Y label ✅ → title ✅
# 🎉 SBA TASK #8 COMPLETE
```

```python
# SBA TASK #9 - Multi-Line Plot
# Phase 1 ==> Make it work.
# 🧠 Carolyn's Self-Talk:
# "This is a LINE plot, so I know I need plt.plot()."
# "month_number gives me my X-axis."
# "Each PRODUCT SALES column will become its own Y-axis line."
# "Each plt.plot() adds another product line to the SAME graph."
df.columns
plt.plot(df['month_number'], df['facecream'])
plt.plot(df['month_number'], df['facewash'])
plt.plot(df['month_number'], df['toothpaste'])
plt.plot(df['month_number'], df['bathingsoap'])
plt.plot(df['month_number'], df['shampoo'])
plt.plot(df['month_number'], df['moisturizer'])
# Then we can say:
# 🧠 “Six products = six plt.plot() commands = one multi-line graph.”

# And THAT finishes Phase 1. Then we'll move to Phase 2.
```

# Phase 2 → Make the graph understandable

```python
plt.plot(df['month_number'], df['facecream'],
         label='Face cream Sales Data')
plt.plot(df['month_number'], df['facewash'],
         label='Face Wash Sales Data')

plt.plot(df['month_number'], df['toothpaste'],
         label='Toothpaste Sales Data')

plt.plot(df['month_number'], df['bathingsoap'],
         label='Bathingsoap Sales Data')

plt.plot(df['month_number'], df['shampoo'],
         label='Shampoo Sales Data')

plt.plot(df['month_number'], df['moisturizer'],
         label='Moisturizer Sales Data')
plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.title('Sales Data')
plt.legend()
```
